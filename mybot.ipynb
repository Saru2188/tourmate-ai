```python
import ollama
```

```python
response = ollama.chat(
    model = 'mistral',
    messages=[
        {
            'role' : 'user',
            'content' : 'what is deep learning'
        }
    ]
)
print(response['message']['content'])
```

```python
user_input = input("ASK A QUESTION")

response = ollama.chat(
    model = 'mistral',
    messages=[
        {
            'role' : 'user',
            'content' : user_input
        }
    ]
)
print(response['message']['content'])
```
